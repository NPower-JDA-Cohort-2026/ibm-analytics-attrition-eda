# Data Wrangling – HR Employee Attrition Dataset

The objective of this notebook is to transform the cleaned and normalized HR Employee Attrition dataset into a structured and analysis-ready format.

Data wrangling involves organizing, transforming, and creating meaningful variables from the existing data to support further analysis.


```python
# Import Libraries and Load Data
import pandas as pd
import numpy as np
df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

# Select relevant columns
Selected the variables that are most relevant to understanding employee attrition and created a separate analysis dataset.

```python
attrition_df = df[
    [
        'Attrition',
        'Age',
        'Gender',
        'Department',
        'JobRole',
        'JobLevel',
        'MonthlyIncome',
        'OverTime',
        'JobSatisfaction',
        'WorkLifeBalance',
        'TotalWorkingYears',
        'YearsAtCompany',
        'YearsSinceLastPromotion',
        'DistanceFromHome'
    ]
].copy()
```

```python
# Converted Yes and No into 1 and 0 so these binary variables can be easily used in calculations.
attrition_df['Attrition'] = attrition_df['Attrition'].map({
    'Yes': 1,
    'No': 0
})

attrition_df['OverTime'] = attrition_df['OverTime'].map({
    'Yes': 1,
    'No': 0
})
```

# Create age and tenure groups to make it easier to compare employee attrition across different employee groups.

```python
# Age Group
attrition_df['AgeGroup'] = pd.cut(
    attrition_df['Age'],
    bins=[0, 30, 40, 50, 100],
    labels=['Under 30', '30-39', '40-49', '50+']
)
```

```python
#Tenure Group
attrition_df['TenureGroup'] = pd.cut(
    attrition_df['YearsAtCompany'],
    bins=[-1, 2, 5, 10, 100],
    labels=['0-2 Years', '3-5 Years', '6-10 Years', '10+ Years']
)
```

```python
# Grouped the data by department and calculated the total employees, employees who left, and attrition rate.
department_summary = (
    attrition_df
    .groupby('Department')
    .agg(
        TotalEmployees=('Attrition', 'count'),
        EmployeesLeft=('Attrition', 'sum'),
        AttritionRate=('Attrition', 'mean')
    )
    .reset_index()
)

department_summary['AttritionRate'] = (
    department_summary['AttritionRate'] * 100
).round(2)

department_summary
```
