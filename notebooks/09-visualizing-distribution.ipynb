## Objective

This notebook uses distribution visualizations to understand how important
numerical employee characteristics are spread across the dataset. The analysis
focuses on age, monthly income, years at the company, and total working years.

```python
# Importing the required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Applying a consistent chart style
sns.set_theme(style="whitegrid")
```

```python
# Loading the IBM employee attrition dataset
df = pd.read_csv(
    "../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

print(f"Dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")

df.head()
```

## Distribution of Employee Age

A histogram is used to show how employee ages are distributed. The curve
shows the overall shape of the distribution, while the vertical lines
identify the mean and median ages.

```python
# Calculating the mean and median employee age
mean_age = df["Age"].mean()
median_age = df["Age"].median()

# Creating the employee age distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True,
    color="#4C78A8",
    edgecolor="white"
)

plt.axvline(
    mean_age,
    color="red",
    linestyle="--",
    label=f"Mean: {mean_age:.1f}"
)

plt.axvline(
    median_age,
    color="green",
    linestyle=":",
    label=f"Median: {median_age:.1f}"
)

plt.title("Distribution of Employee Age")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.legend()
plt.tight_layout()
plt.show()
```

### Age Distribution Finding

Employee ages range from approximately 18 to 60 years, with the highest
concentration occurring in the early-to-late 30s. The mean age is 36.9,
while the median is 36.0. Because the mean is slightly higher than the
median, the distribution has a mild right skew toward older employees.

## Distribution of Monthly Income

This histogram shows how monthly income is distributed among employees.
The mean and median help determine whether the income distribution is
balanced or skewed by higher-income employees.

```python
# Calculating the mean and median monthly income
mean_income = df["MonthlyIncome"].mean()
median_income = df["MonthlyIncome"].median()

# Creating the monthly income distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="MonthlyIncome",
    bins=20,
    kde=True,
    color="#F28E2B",
    edgecolor="white"
)

plt.axvline(
    mean_income,
    color="red",
    linestyle="--",
    label=f"Mean: ${mean_income:,.0f}"
)

plt.axvline(
    median_income,
    color="green",
    linestyle=":",
    label=f"Median: ${median_income:,.0f}"
)

plt.title("Distribution of Employee Monthly Income")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")
plt.legend()
plt.tight_layout()
plt.show()
```

### Monthly Income Distribution Finding

Monthly income is strongly right-skewed. Most employees earn in the
lower-to-middle income range, while a smaller group of high-income
employees creates a long tail toward the right. The mean monthly income
is $6,503, which is considerably higher than the median of $4,919.
Therefore, the median better represents the typical employee's monthly
income.

## Distribution of Years at the Company

This histogram examines employee tenure and shows whether most employees
are relatively new or have remained with the company for many years.

```python
# Calculating the mean and median years at the company
mean_tenure = df["YearsAtCompany"].mean()
median_tenure = df["YearsAtCompany"].median()

# Creating the tenure distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="YearsAtCompany",
    bins=20,
    kde=True,
    color="#59A14F",
    edgecolor="white"
)

plt.axvline(
    mean_tenure,
    color="red",
    linestyle="--",
    label=f"Mean: {mean_tenure:.1f} years"
)

plt.axvline(
    median_tenure,
    color="blue",
    linestyle=":",
    label=f"Median: {median_tenure:.1f} years"
)

plt.title("Distribution of Years at the Company")
plt.xlabel("Years at Company")
plt.ylabel("Number of Employees")
plt.legend()
plt.tight_layout()
plt.show()
```

### Years at the Company Distribution Finding

Employee tenure is strongly right-skewed. Most employees have worked at
the company for approximately 2 to 10 years, while relatively few have
remained for more than 20 years. The mean tenure is 7 years and the
median is 5 years, showing that long-serving employees pull the average
upward.

## Distribution of Total Working Years

This visualization shows the distribution of employees' total professional
experience across their careers, not only their time at the current company.

```python
# Calculating the mean and median total working years
mean_experience = df["TotalWorkingYears"].mean()
median_experience = df["TotalWorkingYears"].median()

# Creating the total working years distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="TotalWorkingYears",
    bins=20,
    kde=True,
    color="#B07AA1",
    edgecolor="white"
)

plt.axvline(
    mean_experience,
    color="red",
    linestyle="--",
    label=f"Mean: {mean_experience:.1f} years"
)

plt.axvline(
    median_experience,
    color="green",
    linestyle=":",
    label=f"Median: {median_experience:.1f} years"
)

plt.title("Distribution of Employees' Total Working Years")
plt.xlabel("Total Working Years")
plt.ylabel("Number of Employees")
plt.legend()
plt.tight_layout()
plt.show()
```

### Total Working Years Distribution Finding

Total working experience is right-skewed. Most employees have approximately
5 to 12 years of total professional experience, while a smaller number have
worked for 20 to 40 years. The mean is 11.3 years and the median is 10 years,
indicating that highly experienced employees pull the average upward.

## Key Findings

- Employee age is approximately balanced with a mild right skew. The mean
  age is 36.9 and the median is 36.
- Monthly income is strongly right-skewed. A relatively small group of
  high-income employees raises the mean above the median.
- Years at the company is also right-skewed, with most employees having
  approximately 2 to 10 years of tenure.
- Total working experience is concentrated around 5 to 12 years, with
  fewer highly experienced employees creating a long right tail.
- For the strongly skewed variables, the median provides a more
  representative measure of a typical employee than the mean.

These visualizations describe the overall employee population. They do
not, by themselves, establish which factors cause employee attrition.
