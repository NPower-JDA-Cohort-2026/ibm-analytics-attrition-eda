```python
import pandas as pd

df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")

df.head()
```

```python
print(df['BusinessTravel'].unique())
```

```python
#Encodes travel frequencies
#0 = Non-Traveler
#1 = Travel_Rarely
#2 = Travel_Frequently
travel_map = {
    'Non-Travel': 0,
    'Travel_Rarely': 1,
    'Travel_Frequently': 2
}
df['BusinessTravel'] = df['BusinessTravel'].map(travel_map)
print(df['BusinessTravel'].unique())
df.head()
```

```python
yn_map = {
    'No': 0,
    'Yes': 1
}
df['Attrition'] = df['Attrition'].map(yn_map)
print(df['Attrition'].unique())
df.head()
```

```python
df['OverTime'] = df['OverTime'].map(yn_map)
print(df['OverTime'].unique())
df.head()
```

```python
df = pd.get_dummies(df, columns=['Gender'], dtype=int)
df.head()
```

```python
#Removed columns that had no predictive variablity (i.e. only one unique value)
df_summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Unique': df.nunique()
})
df_summary
```
