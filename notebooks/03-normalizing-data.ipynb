```python
import pandas as pd

df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")

df.head()
```

```python
#Encodes travel frequencies
#0 = Non-Traveler
#1 = Travel_Rarely
#2 = Travel_Frequently
travel_map = {
    'Non-Travel': 0,
    'Travel_Rarely': 1,
    'Travel_Frequently': 2
}
df['BusinessTravel'] = df['BusinessTravel'].map(travel_map)
print(df['BusinessTravel'].unique())
```

```python
yn_map = {
    'No': 0,
    'Yes': 1
}
df['Attrition'] = df['Attrition'].map(yn_map)
print(df['Attrition'].unique())
```

```python
#Maps yes/no into numeric values. Yes: 1, No: 0
df['OverTime'] = df['OverTime'].map(yn_map)
print(df['OverTime'].unique())
```

```python
#Splits gender into the available categories
df = pd.get_dummies(df, columns=['Gender'], dtype=int)
df_summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Unique': df.nunique()
})
df_summary
```

```python
#Splits marital status into available categories
print(df['MaritalStatus'].unique())
df = pd.get_dummies(df, columns=['MaritalStatus'], dtype=int)
df_summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Unique': df.nunique()
})
df_summary
```

```python
#There is only two numbers for peformance rating.
#Converting them to only one and zero seems like a logical approach

df['PerformanceRating'] = df['PerformanceRating'].map({3:0, 4:1})
df['PerformanceRating']
```
