```python
import pandas as pd

df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")

df.head()
```

```python
yn_map = {
    'No': 0,
    'Yes': 1
}
df['Attrition'] = df['Attrition'].map(yn_map)
print(df['Attrition'].unique())
```

```python
#Maps yes/no into numeric values. Yes: 1, No: 0
df['OverTime'] = df['OverTime'].map(yn_map)
print(df['OverTime'].unique())
```

```python
#Splits gender into the available categories
df = pd.get_dummies(df, 
        columns=['BusinessTravel',
        'Department',
        'EducationField',
        'Gender',
        'JobRole',
        'MaritalStatus'],
        dtype=int)
df_summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Unique': df.nunique()
})
df_summary
```

```python
df[['MonthlyIncome', 'MonthlyRate','HourlyRate']].describe()
```

```python
#There is only two numbers for peformance rating.
#Converting them to only one and zero seems like a logical approach
df['PerformanceRating'] = df['PerformanceRating'].map({3:0, 4:1})
df['PerformanceRating']
```

```python
from sklearn.preprocessing import StandardScaler

numeric_cols = [
    'Age',
    'DailyRate',
    'DistanceFromHome',
    'HourlyRate',
    'MonthlyIncome',
    'MonthlyRate',
    'NumCompaniesWorked',
    'PercentSalaryHike',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

scaler = StandardScaler()

df[[numeric_cols]] = scaler.fit_transform(df[numeric_cols])
df.describe()
```
