```python
# Importing required libraries
import pandas as pd
```

```python
# Loading our attrition data into a pandas dataframe
df = pd.read_csv(
    "../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

# Visualizing top 5 records
df.head()
```

```python
# Check for completely duplicated rows
duplicate_rows = df.duplicated().sum()
print("Number of duplicated rows:", duplicate_rows)
```

```python
# Check duplicates using the EmployerNumber column (Unique Identifier)
duplicate_employee_numbers = df["EmployeeNumber"].duplicated().sum()
print("Duplicated EmployeeNumbers:", duplicate_employee_numbers)
```

```python
# Summary
print("Total rows:", len(df))
print("Completely duplicated rows:", df.duplicated().sum())
print("Unique EmployeeNumbers:", df["EmployeeNumber"].nunique())
print("Duplicated EmployeeNumbers:", df["EmployeeNumber"].duplicated().sum())
```

#### **Conclusion**

- The duplicate analysis confirms that the dataset contains **1,470 records with no completely duplicated rows**. In addition, all **1,470 EmployeeNumbers are unique**, with no duplicated employee identifiers identified. These findings indicate that each record represents a distinct employee and that there are no apparent duplicate records requiring removal based on the checks performed. The dataset can therefore proceed to the next stage of data preparation and analysis.
